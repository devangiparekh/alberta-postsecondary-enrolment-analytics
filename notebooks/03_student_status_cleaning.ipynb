## Reshape Student Status Data

The Statistics Canada extract stores total, Canadian, and international
enrolment in three separate 10-year column blocks.

Each block is reshaped into long format and assigned a student-status
category so that each record represents one institution, one academic
year, and one student-status group.

```python
def reshape_status_block(data, columns, status_name):
    
    block = data[
        ["Institution_Raw", "Institution", "Institution_Standardized"] + columns
    ].copy()

    block = block.melt(
        id_vars=[
            "Institution_Raw",
            "Institution",
            "Institution_Standardized"
        ],
        value_vars=columns,
        var_name="Academic_Year",
        value_name="Enrolment"
    )

    block["Student_Status"] = status_name

    return block
```

```python
df_total = reshape_status_block(
    df,
    total_cols,
    "Total"
)

df_canadian = reshape_status_block(
    df,
    canadian_cols,
    "Canadian"
)

df_international = reshape_status_block(
    df,
    international_cols,
    "International"
)
```

```python
print("Total:", df_total.shape)
print("Canadian:", df_canadian.shape)
print("International:", df_international.shape)
```

```python
for block in [df_total, df_canadian, df_international]:
    block["Academic_Year"] = (
        block["Academic_Year"]
        .str.replace(r"\.\d+$", "", regex=True)
        .str.replace(" ", "", regex=False)
    )
```

```python
print(df_total["Academic_Year"].unique())
print(df_canadian["Academic_Year"].unique())
print(df_international["Academic_Year"].unique())
```

```python
df_long = pd.concat(
    [
        df_total,
        df_canadian,
        df_international
    ],
    ignore_index=True
)

df_long.shape
```

```python
df_long["Student_Status"].value_counts()
```

```python
df_long["Enrolment"] = (
    df_long["Enrolment"]
    .astype(str)
    .str.replace(",", "", regex=False)
)

df_long["Enrolment"] = pd.to_numeric(
    df_long["Enrolment"],
    errors="coerce"
)
```

```python
df_long = df_long[
    ~(
        (
            df_long["Institution_Standardized"]
            == "Alberta University of the Arts"
        )
        &
        (
            df_long["Enrolment"].isna()
        )
    )
].copy()

df_long = df_long.reset_index(drop=True)
```

```python
df_long.shape
```

```python
df_long["Province"] = "Alberta"

df_long["Record_Type"] = df_long[
    "Institution_Standardized"
].apply(
    lambda x: "Province Total"
    if x == "Alberta"
    else "Institution"
)
```

```python
df_long["Academic_Year_Start"] = (
    df_long["Academic_Year"]
    .str[:4]
    .astype(int)
)

df_long["Academic_Year_End"] = (
    df_long["Academic_Year_Start"] + 1
)
```

```python
validation = df_long.pivot_table(
    index=[
        "Institution_Standardized",
        "Academic_Year"
    ],
    columns="Student_Status",
    values="Enrolment",
    aggfunc="sum"
).reset_index()
```

```python
validation.head()
```

```python
validation["Calculated_Total"] = (
    validation["Canadian"]
    + validation["International"]
)
```

```python
validation["Difference"] = (
    validation["Total"]
    - validation["Calculated_Total"]
)
```

```python
print("Number of validation records:", len(validation))

print(
    "Exact matches:",
    (validation["Difference"] == 0).sum()
)

print(
    "Within ±3:",
    (validation["Difference"].abs() <= 3).sum()
)

print(
    "Within ±10:",
    (validation["Difference"].abs() <= 10).sum()
)

print(
    "Largest absolute difference:",
    validation["Difference"].abs().max()
)

print(
    "Mean absolute difference:",
    validation["Difference"].abs().mean()
)
```

```python
validation[
    [
        "Institution_Standardized",
        "Academic_Year",
        "Total",
        "Canadian",
        "International",
        "Difference"
    ]
].assign(
    Abs_Difference=lambda x: x["Difference"].abs()
).sort_values(
    "Abs_Difference",
    ascending=False
).head(15)
```

```python
validation["Difference_Pct"] = (
    validation["Difference"].abs()
    / validation["Total"]
    * 100
)

validation[
    [
        "Institution_Standardized",
        "Academic_Year",
        "Total",
        "Canadian",
        "International",
        "Difference",
        "Difference_Pct"
    ]
].sort_values(
    "Difference_Pct",
    ascending=False
).head(15)
```

```python
print(
    "Maximum percentage discrepancy:",
    round(validation["Difference_Pct"].max(), 3),
    "%"
)

print(
    "Mean percentage discrepancy:",
    round(validation["Difference_Pct"].mean(), 3),
    "%"
)

print(
    "Records within 0.5%:",
    (validation["Difference_Pct"] <= 0.5).sum(),
    "of",
    len(validation)
)

print(
    "Records within 1%:",
    (validation["Difference_Pct"] <= 1).sum(),
    "of",
    len(validation)
)
```

## Final Data Quality Check

Before saving the cleaned student-status dataset, perform final checks for:

- Missing enrolment values
- Duplicate institution-year-status records
- Negative enrolment values
- Number of institutions
- Number of academic years
- Number of records by student status

```python
print("FINAL DATA QUALITY CHECK — DATASET 2")
print("-" * 45)

print("Rows:", len(df_long))
print("Columns:", df_long.shape[1])

print(
    "Institutions:",
    df_long["Institution_Standardized"].nunique()
)

print(
    "Academic years:",
    df_long["Academic_Year"].nunique()
)

print(
    "Missing enrolment values:",
    df_long["Enrolment"].isna().sum()
)

print(
    "Negative enrolment values:",
    (df_long["Enrolment"] < 0).sum()
)

print(
    "Duplicate institution-year-status records:",
    df_long.duplicated(
        subset=[
            "Institution_Standardized",
            "Academic_Year",
            "Student_Status"
        ]
    ).sum()
)

print("\nRecords by student status:")
print(df_long["Student_Status"].value_counts())
```
