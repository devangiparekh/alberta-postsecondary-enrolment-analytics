## 01 - Data Cleaning and Preparation

### Objective

Prepare Statistics Canada post-secondary enrolment data for analysis of historical enrolment trends across Alberta institutions from 2014/15 through 2023/24.

### Data Source

Statistics Canada — Postsecondary enrolment data.

### Cleaning Goals

- Preserve the original source data
- Inspect the dataset structure
- Identify missing or inconsistent values
- Standardize institution and academic-year fields
- Convert enrolment values into an analysis-ready format
- Validate cleaned values against the original source
- Export a clean dataset for subsequent analysis

```python
last_data_row = df[
    df["Geography and institutions"].eq("Portage College, Alberta")
].index[0]
```

```python
df = df.rename(
    columns={
        "Geography and institutions": "Institution_Raw"
    }
)
```

```python
df["Institution"] = (
    df["Institution_Raw"]
    .str.replace(r"\s+11$", "", regex=True)
    .str.replace(r", Alberta$", "", regex=True)
    .str.strip()
)
```

```python
df[["Institution_Raw", "Institution"]].head(15)
```

```python
year_columns = [
    "2014 / 2015",
    "2015 / 2016",
    "2016 / 2017",
    "2017 / 2018",
    "2018 / 2019",
    "2019 / 2020",
    "2020 / 2021",
    "2021 / 2022",
    "2022 / 2023",
    "2023 / 2024"
]
```

```python
for col in year_columns:
    df[col] = (
        df[col]
        .str.replace(",", "", regex=False)
        .pipe(pd.to_numeric, errors="coerce")
    )
```

```python
df_long["Academic_Year"] = (
    df_long["Academic_Year"]
    .str.replace(" ", "", regex=False)
)
```

```python
df_long["Academic_Year"].unique()
```

```python
df_long["Academic_Year_Start"] = (
    df_long["Academic_Year"]
    .str[:4]
    .astype(int)
)
```

```python
df_long["Academic_Year_End"] = (
    df_long["Academic_Year_Start"] + 1
)
```

```python
df_long[
    ["Academic_Year", "Academic_Year_Start", "Academic_Year_End"]
].drop_duplicates()
```

```python
df_long["Province"] = "Alberta"
```

```python
df_long["Record_Type"] = df_long["Institution"].apply(
    lambda x: "Province Total" if x == "Alberta" else "Institution"
)
```

## Historical Institution Name Harmonization

During data-quality review, Alberta College of Art and Design and
Alberta University of the Arts appeared as separate Statistics Canada
records with mutually exclusive reporting periods.

The historical and current names were standardized under
"Alberta University of the Arts" for longitudinal analysis while the
original source institution name was retained for traceability.

Rows associated with the inactive institution name and containing no
reported enrolment value were removed rather than interpreted as zero
enrolment.

```python
alberta_total = (
    df_long[
        df_long["Record_Type"] == "Province Total"
    ][
        ["Academic_Year", "Academic_Year_Start", "Enrolment"]
    ]
    .rename(columns={"Enrolment": "Reported_Alberta_Total"})
    .sort_values("Academic_Year_Start")
)

alberta_total
```

```python
institution_sum = (
    df_long[
        df_long["Record_Type"] == "Institution"
    ]
    .groupby(
        ["Academic_Year", "Academic_Year_Start"],
        as_index=False
    )["Enrolment"]
    .sum()
    .rename(columns={"Enrolment": "Institution_Sum"})
    .sort_values("Academic_Year_Start")
)

institution_sum
```

```python
reconciliation = alberta_total.merge(
    institution_sum,
    on=["Academic_Year", "Academic_Year_Start"],
    how="left"
)
```

```python
reconciliation["Difference"] = (
    reconciliation["Reported_Alberta_Total"]
    - reconciliation["Institution_Sum"]
)
```

```python
reconciliation["Difference_Pct"] = (
    reconciliation["Difference"]
    / reconciliation["Reported_Alberta_Total"]
    * 100
)
```

## Provincial Total Reconciliation

As a final validation step(above), the sum of institution-level enrolment
records was compared with Statistics Canada's reported Alberta
provincial total for each academic year.

This test is intended to identify potential differences in coverage,
classification, aggregation, or reporting before the dataset is used
for analytical modelling.

### Reconciliation Result

Institution-level enrolment totals were compared with the published
Alberta provincial totals for all 10 academic years.

Differences ranged from 0 to 12 students annually, representing less
than 0.01% of reported provincial enrolment in every year.

These minor discrepancies are consistent with Statistics Canada's
random-rounding methodology. Published values were therefore retained
without adjustment.

The reconciliation provides additional confidence that the cleaned
dataset accurately preserves the source data and that no material
records were lost during transformation.
